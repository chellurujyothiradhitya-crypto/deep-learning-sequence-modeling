```python
# ============================================================
# SIMPLE RNN FOR TEMPERATURE TIME-SERIES PREDICTION
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from tensorflow.keras.optimizers import Adam


# ------------------------------------------------------------
# 1. DATASET PATH
# ------------------------------------------------------------

# Change this path according to your Kaggle dataset
DATASET_PATH = "/kaggle/input/datasets/jyothiradhityach/temprature-csv/average-annual-surface-temperature.csv"

# Load dataset
data = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully!")
print("Dataset shape:", data.shape)

# Display first few records
print("\nFirst 5 records:")
print(data.head())


# ------------------------------------------------------------
# 2. SELECT TEMPERATURE COLUMN
# ------------------------------------------------------------

# CSV file should contain a column named "Temperature"

if "Average surface temperature" not in data.columns:
    raise ValueError(
        "Column 'Temperature' not found in the dataset.\n"
        f"Available columns: {list(data.columns)}"
    )

temperature = data["Average surface temperature"].values.reshape(-1, 1)


# ------------------------------------------------------------
# 3. HANDLE MISSING VALUES
# ------------------------------------------------------------

temperature_df = pd.DataFrame(temperature)

temperature_df = temperature_df.interpolate(
    method="linear"
)

# Handle missing values at beginning/end
temperature_df = temperature_df.bfill().ffill()

temperature = temperature_df.values


print("\nTemperature data:")
print("Number of samples:", len(temperature))
print("Minimum temperature:", temperature.min())
print("Maximum temperature:", temperature.max())


# ------------------------------------------------------------
# 4. NORMALIZATION
# ------------------------------------------------------------

scaler = MinMaxScaler()

temperature_scaled = scaler.fit_transform(temperature)


# ------------------------------------------------------------
# 5. CREATE SEQUENCES USING SLIDING WINDOW
# ------------------------------------------------------------

sequence_length = 20

X = []
y = []

for i in range(len(temperature_scaled) - sequence_length):

    # Previous 20 temperature values
    X.append(
        temperature_scaled[
            i:i + sequence_length
        ]
    )

    # Next temperature value
    y.append(
        temperature_scaled[
            i + sequence_length
        ]
    )


X = np.array(X)
y = np.array(y)

print("\nSequence information")
print("---------------------------")
print("Input shape :", X.shape)
print("Output shape:", y.shape)


# ------------------------------------------------------------
# 6. TRAIN-TEST SPLIT
# ------------------------------------------------------------

# Time-series data should be split chronologically.

train_size = int(len(X) * 0.8)

X_train = X[:train_size]
y_train = y[:train_size]

X_test = X[train_size:]
y_test = y[train_size:]

print("\nDataset Split")
print("---------------------------")
print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))


# ------------------------------------------------------------
# 7. BUILD SIMPLE RNN MODEL
# ------------------------------------------------------------

model = Sequential([

    SimpleRNN(
        units=64,
        activation="tanh",
        input_shape=(sequence_length, 1)
    ),

    Dense(1)
])


# ------------------------------------------------------------
# 8. COMPILE MODEL
# ------------------------------------------------------------

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse"
)

print("\nModel Summary")
print("---------------------------")

model.summary()


# ------------------------------------------------------------
# 9. TRAIN MODEL
# ------------------------------------------------------------

print("\nTraining Simple RNN...")
print("---------------------------")

history = model.fit(
    X_train,
    y_train,

    validation_split=0.2,

    epochs=50,

    batch_size=32,

    verbose=1
)


# ------------------------------------------------------------
# 10. PREDICTION
# ------------------------------------------------------------

print("\nGenerating predictions...")

predicted_scaled = model.predict(X_test)

# Convert normalized values back to original temperature
predicted = scaler.inverse_transform(
    predicted_scaled
)

actual = scaler.inverse_transform(
    y_test
)


# ------------------------------------------------------------
# 11. DISPLAY PREDICTIONS
# ------------------------------------------------------------

print("\nActual vs Predicted Temperature")
print("----------------------------------------")

for i in range(min(10, len(actual))):

    print(
        f"Actual: {actual[i][0]:.2f}   "
        f"Predicted: {predicted[i][0]:.2f}"
    )


# ------------------------------------------------------------
# 12. PLOT TRAINING AND VALIDATION LOSS
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")

plt.title(
    "Simple RNN - Training and Validation Loss"
)

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 13. ACTUAL VS PREDICTED TEMPERATURE
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    actual,
    label="Actual Temperature"
)

plt.plot(
    predicted,
    label="Predicted Temperature"
)

plt.xlabel("Time")
plt.ylabel("Temperature")

plt.title(
    "Simple RNN - Actual vs Predicted Temperature"
)

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 14. EVALUATION METRICS
# ------------------------------------------------------------

mae = mean_absolute_error(
    actual,
    predicted
)

mse = mean_squared_error(
    actual,
    predicted
)

rmse = np.sqrt(mse)


print("\nEvaluation Results")
print("---------------------------")

print("MAE  :", mae)
print("MSE  :", mse)
print("RMSE :", rmse)


# ------------------------------------------------------------
# 15. FINAL MESSAGE
# ------------------------------------------------------------

print("\n========================================")
print("Simple RNN Experiment Completed!")
print("========================================")
```

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, SimpleRNN, Dense
from tensorflow.keras.optimizers import Adam


# ============================================================
# LOAD DATA
# ============================================================

DATASET_PATH = "/kaggle/input/datasets/jyothiradhityach/temprature-csv/average-annual-surface-temperature.csv"

data = pd.read_csv(DATASET_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", data.shape)
print("\nColumns:")
print(data.columns.tolist())

# Select temperature column
temperature = data[
    "Average surface temperature"
].values.reshape(-1, 1)


# ============================================================
# HANDLE MISSING VALUES
# ============================================================

temperature = pd.DataFrame(
    temperature
).interpolate().bfill().ffill().values


# ============================================================
# NORMALIZATION
# ============================================================

scaler = MinMaxScaler()

temperature_scaled = scaler.fit_transform(
    temperature
)


# ============================================================
# CREATE SEQUENCES
# ============================================================

def create_sequences(data, sequence_length):

    X = []
    y = []

    for i in range(
        len(data) - sequence_length
    ):

        X.append(
            data[
                i:i + sequence_length
            ]
        )

        y.append(
            data[
                i + sequence_length
            ]
        )

    return np.array(X), np.array(y)


# ============================================================
# TRAIN AND EVALUATE RNN
# ============================================================

def train_rnn(
    sequence_length,
    hidden_units,
    learning_rate,
    batch_size,
    epochs
):

    # Create sequences
    X, y = create_sequences(
        temperature_scaled,
        sequence_length
    )

    # Chronological train-test split
    train_size = int(
        len(X) * 0.8
    )

    X_train = X[:train_size]
    y_train = y[:train_size]

    X_test = X[train_size:]
    y_test = y[train_size:]


    # --------------------------------------------------------
    # BUILD MODEL
    # --------------------------------------------------------

    model = Sequential([

        # NEW KERAS RECOMMENDED METHOD
        Input(
            shape=(
                sequence_length,
                1
            )
        ),

        SimpleRNN(
            hidden_units,
            activation="tanh"
        ),

        Dense(1)
    ])


    # --------------------------------------------------------
    # COMPILE
    # --------------------------------------------------------

    model.compile(

        optimizer=Adam(
            learning_rate=learning_rate
        ),

        loss="mse"
    )


    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    model.fit(

        X_train,
        y_train,

        epochs=epochs,

        batch_size=batch_size,

        validation_split=0.2,

        verbose=0
    )


    # --------------------------------------------------------
    # PREDICTION
    # --------------------------------------------------------

    prediction_scaled = model.predict(
        X_test,
        verbose=0
    )


    # Convert back to original temperature
    prediction = scaler.inverse_transform(
        prediction_scaled
    )

    actual = scaler.inverse_transform(
        y_test
    )


    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------

    mae = mean_absolute_error(
        actual,
        prediction
    )

    mse = mean_squared_error(
        actual,
        prediction
    )

    rmse = np.sqrt(mse)


    return mae, mse, rmse


# ============================================================
# EXPERIMENTS
# ============================================================

experiments = [

    # Sequence Length
    ("Sequence Length", 10, 64, 0.001, 32, 50),
    ("Sequence Length", 20, 64, 0.001, 32, 50),
    ("Sequence Length", 30, 64, 0.001, 32, 50),

    # Hidden Units
    ("Hidden Units", 20, 32, 0.001, 32, 50),
    ("Hidden Units", 20, 64, 0.001, 32, 50),
    ("Hidden Units", 20, 128, 0.001, 32, 50),

    # Learning Rate
    ("Learning Rate", 20, 64, 0.0001, 32, 50),
    ("Learning Rate", 20, 64, 0.001, 32, 50),
    ("Learning Rate", 20, 64, 0.01, 32, 50),

    # Batch Size
    ("Batch Size", 20, 64, 0.001, 16, 50),
    ("Batch Size", 20, 64, 0.001, 32, 50),
    ("Batch Size", 20, 64, 0.001, 64, 50),

    # Epochs
    ("Epochs", 20, 64, 0.001, 32, 25),
    ("Epochs", 20, 64, 0.001, 32, 50),
    ("Epochs", 20, 64, 0.001, 32, 100)
]


# ============================================================
# RUN EXPERIMENTS
# ============================================================

results = []

print("\nStarting experiments...")
print("=" * 70)

for experiment in experiments:

    parameter = experiment[0]

    sequence_length = experiment[1]
    hidden_units = experiment[2]
    learning_rate = experiment[3]
    batch_size = experiment[4]
    epochs = experiment[5]


    print(
        f"\nRunning: {parameter} | "
        f"Seq={sequence_length} | "
        f"Units={hidden_units} | "
        f"LR={learning_rate} | "
        f"Batch={batch_size} | "
        f"Epochs={epochs}"
    )


    mae, mse, rmse = train_rnn(

        sequence_length,

        hidden_units,

        learning_rate,

        batch_size,

        epochs
    )


    results.append([

        parameter,

        sequence_length,

        hidden_units,

        learning_rate,

        batch_size,

        epochs,

        mae,

        mse,

        rmse
    ])


# ============================================================
# RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(

    results,

    columns=[

        "Parameter",

        "Sequence Length",

        "Hidden Units",

        "Learning Rate",

        "Batch Size",

        "Epochs",

        "MAE",

        "MSE",

        "RMSE"
    ]
)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("\n\nFINAL EXPERIMENT RESULTS")
print("=" * 100)

print(
    results_df.to_string(
        index=False
    )
)


# ============================================================
# FIND BEST MODEL
# ============================================================

best_result = results_df.loc[
    results_df["RMSE"].idxmin()
]

print("\n\nBEST CONFIGURATION")
print("=" * 50)

print(
    best_result
)
```

```python
# ============================================================
# Q9 - LSTM FOR TEMPERATURE TIME-SERIES PREDICTION
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.optimizers import Adam


# ------------------------------------------------------------
# 1. DATASET PATH
# ------------------------------------------------------------

DATASET_PATH = "/kaggle/input/datasets/jyothiradhityach/temprature-csv/average-annual-surface-temperature.csv"

# Load dataset
data = pd.read_csv(DATASET_PATH)

print("\nDataset loaded successfully!")
print("Dataset shape:", data.shape)

print("\nFirst 5 records:")
print(data.head())


# ------------------------------------------------------------
# 2. SELECT TEMPERATURE COLUMN
# ------------------------------------------------------------

if "Average surface temperature" not in data.columns:
    raise ValueError(
        "Column 'Average surface temperature' not found.\n"
        f"Available columns: {list(data.columns)}"
    )

temperature = data[
    "Average surface temperature"
].values.reshape(-1, 1)


# ------------------------------------------------------------
# 3. HANDLE MISSING VALUES
# ------------------------------------------------------------

temperature_df = pd.DataFrame(temperature)

temperature_df = temperature_df.interpolate(
    method="linear"
)

temperature_df = temperature_df.bfill().ffill()

temperature = temperature_df.values


print("\nTemperature data:")
print("Number of samples:", len(temperature))
print("Minimum temperature:", temperature.min())
print("Maximum temperature:", temperature.max())


# ------------------------------------------------------------
# 4. NORMALIZATION
# ------------------------------------------------------------

scaler = MinMaxScaler()

temperature_scaled = scaler.fit_transform(
    temperature
)


# ------------------------------------------------------------
# 5. CREATE SEQUENCES
# ------------------------------------------------------------

sequence_length = 20

X = []
y = []

for i in range(
    len(temperature_scaled) - sequence_length
):

    # Previous 20 temperature values
    X.append(
        temperature_scaled[
            i:i + sequence_length
        ]
    )

    # Next temperature value
    y.append(
        temperature_scaled[
            i + sequence_length
        ]
    )


X = np.array(X)
y = np.array(y)


print("\nSequence information")
print("---------------------------")
print("Input shape :", X.shape)
print("Output shape:", y.shape)


# ------------------------------------------------------------
# 6. TRAIN-TEST SPLIT
# ------------------------------------------------------------

# Same chronological 80/20 split as Part A

train_size = int(len(X) * 0.8)

X_train = X[:train_size]
y_train = y[:train_size]

X_test = X[train_size:]
y_test = y[train_size:]


print("\nDataset Split")
print("---------------------------")
print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))


# ------------------------------------------------------------
# 7. BUILD LSTM MODEL
# ------------------------------------------------------------

lstm_model = Sequential([

    LSTM(
        units=64,
        activation="tanh",
        input_shape=(sequence_length, 1)
    ),

    Dense(1)
])


# ------------------------------------------------------------
# 8. COMPILE MODEL
# ------------------------------------------------------------

lstm_model.compile(
    optimizer=Adam(
        learning_rate=0.001
    ),
    loss="mse"
)


print("\nLSTM Model Summary")
print("---------------------------")

lstm_model.summary()


# ------------------------------------------------------------
# 9. TRAIN LSTM MODEL
# ------------------------------------------------------------

print("\nTraining LSTM...")
print("---------------------------")

start_time = time.time()

lstm_history = lstm_model.fit(

    X_train,
    y_train,

    validation_split=0.2,

    epochs=50,

    batch_size=32,

    verbose=1
)

end_time = time.time()

lstm_training_time = end_time - start_time


# ------------------------------------------------------------
# 10. PREDICTION
# ------------------------------------------------------------

print("\nGenerating LSTM predictions...")

lstm_predicted_scaled = lstm_model.predict(
    X_test
)

# Convert normalized values back
# to original temperature scale

lstm_predicted = scaler.inverse_transform(
    lstm_predicted_scaled
)

lstm_actual = scaler.inverse_transform(
    y_test
)


# ------------------------------------------------------------
# 11. DISPLAY PREDICTIONS
# ------------------------------------------------------------

print("\nLSTM - Actual vs Predicted Temperature")
print("----------------------------------------")

for i in range(
    min(10, len(lstm_actual))
):

    print(
        f"Actual: {lstm_actual[i][0]:.2f}   "
        f"Predicted: {lstm_predicted[i][0]:.2f}"
    )


# ------------------------------------------------------------
# 12. TRAINING AND VALIDATION LOSS
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    lstm_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    lstm_history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")

plt.title(
    "LSTM - Training and Validation Loss"
)

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 13. ACTUAL VS PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    lstm_actual,
    label="Actual Temperature"
)

plt.plot(
    lstm_predicted,
    label="LSTM Predicted Temperature"
)

plt.xlabel("Time")
plt.ylabel("Temperature")

plt.title(
    "LSTM - Actual vs Predicted Temperature"
)

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 14. EVALUATION METRICS
# ------------------------------------------------------------

lstm_mae = mean_absolute_error(
    lstm_actual,
    lstm_predicted
)

lstm_mse = mean_squared_error(
    lstm_actual,
    lstm_predicted
)

lstm_rmse = np.sqrt(
    lstm_mse
)


# ------------------------------------------------------------
# 15. MODEL PARAMETERS
# ------------------------------------------------------------

lstm_parameters = lstm_model.count_params()


# ------------------------------------------------------------
# 16. FINAL RESULTS
# ------------------------------------------------------------

print("\n========================================")
print("Q9 - LSTM EXPERIMENT RESULTS")
print("========================================")

print(f"MAE              : {lstm_mae:.6f}")
print(f"MSE              : {lstm_mse:.6f}")
print(f"RMSE             : {lstm_rmse:.6f}")
print(f"Trainable Params : {lstm_parameters:,}")
print(f"Training Time    : {lstm_training_time:.2f} seconds")

print("\n========================================")
print("LSTM Experiment Completed!")
print("========================================")
```

```python
# ============================================================
# Q12 - GRU BASED TEMPERATURE PREDICTION
# Same dataset and preprocessing as Part A and Q9
# ============================================================

import numpy as np
import pandas as pd
import time
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense
from tensorflow.keras.optimizers import Adam


# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

data = pd.read_csv(
    "/kaggle/input/datasets/jyothiradhityach/temprature-csv/average-annual-surface-temperature.csv"
)

print("Dataset loaded successfully!")
print("Dataset shape:", data.shape)


# ------------------------------------------------------------
# SELECT TARGET
# ------------------------------------------------------------

temperature = data["Average surface temperature"].copy()


# ------------------------------------------------------------
# HANDLE MISSING VALUES
# Same preprocessing as Part A
# ------------------------------------------------------------

temperature = temperature.interpolate(method="linear")
temperature = temperature.bfill()
temperature = temperature.ffill()


# ------------------------------------------------------------
# NORMALIZATION
# Same as Part A
# ------------------------------------------------------------

scaler = MinMaxScaler()

temperature_scaled = scaler.fit_transform(
    temperature.values.reshape(-1, 1)
)


# ------------------------------------------------------------
# CREATE SEQUENCES
# Same sequence length as Part A
# ------------------------------------------------------------

sequence_length = 20

X = []
y = []

for i in range(sequence_length, len(temperature_scaled)):
    X.append(
        temperature_scaled[i-sequence_length:i]
    )
    y.append(
        temperature_scaled[i]
    )

X = np.array(X)
y = np.array(y)

print("\nSequence information")
print("---------------------------")
print("X shape:", X.shape)
print("y shape:", y.shape)


# ------------------------------------------------------------
# TRAIN / TEST SPLIT
# Chronological 80/20 split
# Same as Part A
# ------------------------------------------------------------

split_index = int(len(X) * 0.8)

X_train = X[:split_index]
X_test = X[split_index:]

y_train = y[:split_index]
y_test = y[split_index:]

print("\nTraining and testing information")
print("---------------------------")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)


# ------------------------------------------------------------
# BUILD GRU MODEL
# ------------------------------------------------------------

model = Sequential([
    GRU(
        units=64,
        activation="tanh",
        input_shape=(sequence_length, 1)
    ),
    Dense(1)
])


# ------------------------------------------------------------
# COMPILE
# ------------------------------------------------------------

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse"
)


# ------------------------------------------------------------
# DISPLAY MODEL
# ------------------------------------------------------------

print("\n========================================")
print("Q12 - GRU MODEL")
print("========================================")

model.summary()


# ------------------------------------------------------------
# TRAIN MODEL
# Same settings as Part A and Q9
# ------------------------------------------------------------

start_time = time.time()

history = model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)

training_time = time.time() - start_time


# ------------------------------------------------------------
# PREDICTION
# ------------------------------------------------------------

y_pred_scaled = model.predict(X_test, verbose=0)


# ------------------------------------------------------------
# INVERSE TRANSFORM
# Convert normalized values back to temperature
# ------------------------------------------------------------

y_test_actual = scaler.inverse_transform(y_test)
y_pred_actual = scaler.inverse_transform(y_pred_scaled)


# ------------------------------------------------------------
# EVALUATION METRICS
# ------------------------------------------------------------

mae = mean_absolute_error(
    y_test_actual,
    y_pred_actual
)

mse = mean_squared_error(
    y_test_actual,
    y_pred_actual
)

rmse = np.sqrt(mse)


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\n========================================")
print("Q12 - GRU EXPERIMENT RESULTS")
print("========================================")

print(f"MAE              : {mae:.6f}")
print(f"MSE              : {mse:.6f}")
print(f"RMSE             : {rmse:.6f}")
print(f"Trainable Params : {model.count_params():,}")
print(f"Training Time    : {training_time:.2f} seconds")


# ------------------------------------------------------------
# SAMPLE PREDICTIONS
# ------------------------------------------------------------

print("\nFirst 10 Actual vs Predicted Values")
print("----------------------------------------")

for i in range(10):
    print(
        f"Actual: {y_test_actual[i][0]:.2f}  |  "
        f"Predicted: {y_pred_actual[i][0]:.2f}"
    )


# ------------------------------------------------------------
# PLOT 1 - TRAINING AND VALIDATION LOSS
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("GRU Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.grid(True)
plt.show()


# ------------------------------------------------------------
# PLOT 2 - ACTUAL VS PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(12, 5))

plt.plot(
    y_test_actual[:200],
    label="Actual Temperature"
)

plt.plot(
    y_pred_actual[:200],
    label="GRU Predicted Temperature"
)

plt.title("GRU - Actual vs Predicted Temperature")
plt.xlabel("Test Sample")
plt.ylabel("Average Surface Temperature")
plt.legend()
plt.grid(True)
plt.show()


# ------------------------------------------------------------
# FINAL MESSAGE
# ------------------------------------------------------------

print("\n========================================")
print("GRU Experiment Completed!")
print("========================================")
```

```python
# ============================================================
# Q13 - COMPARISON OF RNN, LSTM AND GRU
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Dense
from tensorflow.keras.optimizers import Adam


# ------------------------------------------------------------
# 1. LOAD THE SAME DATASET USED IN PART A
# ------------------------------------------------------------

DATASET_PATH = "/kaggle/input/datasets/jyothiradhityach/temprature-csv/average-annual-surface-temperature.csv"

data = pd.read_csv(DATASET_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", data.shape)


# ------------------------------------------------------------
# 2. SELECT TEMPERATURE COLUMN
# ------------------------------------------------------------

temperature = data["Average surface temperature"].values.reshape(-1, 1)


# ------------------------------------------------------------
# 3. HANDLE MISSING VALUES
#    SAME AS PART A
# ------------------------------------------------------------

temperature_df = pd.DataFrame(temperature)

temperature_df = temperature_df.interpolate(
    method="linear"
)

temperature_df = temperature_df.bfill().ffill()

temperature = temperature_df.values


# ------------------------------------------------------------
# 4. NORMALIZATION
#    SAME AS PART A
# ------------------------------------------------------------

scaler = MinMaxScaler()

temperature_scaled = scaler.fit_transform(
    temperature
)


# ------------------------------------------------------------
# 5. CREATE SEQUENCES
#    SAME AS PART A
# ------------------------------------------------------------

sequence_length = 20

X = []
y = []

for i in range(len(temperature_scaled) - sequence_length):

    X.append(
        temperature_scaled[
            i:i + sequence_length
        ]
    )

    y.append(
        temperature_scaled[
            i + sequence_length
        ]
    )


X = np.array(X)
y = np.array(y)


# ------------------------------------------------------------
# 6. CHRONOLOGICAL TRAIN-TEST SPLIT
#    SAME AS PART A
# ------------------------------------------------------------

train_size = int(len(X) * 0.8)

X_train = X[:train_size]
y_train = y[:train_size]

X_test = X[train_size:]
y_test = y[train_size:]


print("\nDataset Information")
print("--------------------------------")
print("Input shape       :", X.shape)
print("Output shape      :", y.shape)
print("Training samples  :", len(X_train))
print("Testing samples   :", len(X_test))


# ------------------------------------------------------------
# 7. FUNCTION TO TRAIN AND EVALUATE A MODEL
# ------------------------------------------------------------

def train_and_evaluate(model, model_name):

    print("\n" + "=" * 60)
    print(f"Training {model_name}")
    print("=" * 60)

    # Compile
    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss="mse"
    )

    # Training time
    start_time = time.time()

    history = model.fit(
        X_train,
        y_train,
        validation_split=0.2,
        epochs=50,
        batch_size=32,
        verbose=0
    )

    training_time = time.time() - start_time

    # Prediction
    predicted_scaled = model.predict(
        X_test,
        verbose=0
    )

    # Convert back to original temperature scale
    predicted = scaler.inverse_transform(
        predicted_scaled
    )

    actual = scaler.inverse_transform(
        y_test
    )

    # Metrics
    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    # Final losses
    final_train_loss = history.history["loss"][-1]
    final_val_loss = history.history["val_loss"][-1]

    # Best validation loss
    best_val_loss = min(
        history.history["val_loss"]
    )

    # Parameters
    parameters = model.count_params()

    print(f"MAE                : {mae:.6f}")
    print(f"MSE                : {mse:.6f}")
    print(f"RMSE               : {rmse:.6f}")
    print(f"Trainable Params   : {parameters:,}")
    print(f"Training Time      : {training_time:.2f} seconds")
    print(f"Final Train Loss   : {final_train_loss:.6f}")
    print(f"Final Val Loss     : {final_val_loss:.6f}")
    print(f"Best Val Loss      : {best_val_loss:.6f}")

    return {
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "Parameters": parameters,
        "Training Time (sec)": training_time,
        "Final Train Loss": final_train_loss,
        "Final Val Loss": final_val_loss,
        "Best Val Loss": best_val_loss,
        "History": history,
        "Actual": actual,
        "Predicted": predicted
    }


# ------------------------------------------------------------
# 8. BUILD SIMPLE RNN
#    SAME ARCHITECTURE AS PART A
# ------------------------------------------------------------

rnn_model = Sequential([
    SimpleRNN(
        units=64,
        activation="tanh",
        input_shape=(sequence_length, 1)
    ),
    Dense(1)
])


# ------------------------------------------------------------
# 9. BUILD LSTM
#    SAME ARCHITECTURE AS Q9
# ------------------------------------------------------------

lstm_model = Sequential([
    LSTM(
        units=64,
        activation="tanh",
        input_shape=(sequence_length, 1)
    ),
    Dense(1)
])


# ------------------------------------------------------------
# 10. BUILD GRU
#     SAME ARCHITECTURE AS Q12
# ------------------------------------------------------------

gru_model = Sequential([
    GRU(
        units=64,
        activation="tanh",
        input_shape=(sequence_length, 1)
    ),
    Dense(1)
])


# ------------------------------------------------------------
# 11. TRAIN ALL THREE MODELS
# ------------------------------------------------------------

rnn_result = train_and_evaluate(
    rnn_model,
    "Simple RNN"
)

lstm_result = train_and_evaluate(
    lstm_model,
    "LSTM"
)

gru_result = train_and_evaluate(
    gru_model,
    "GRU"
)


# ------------------------------------------------------------
# 12. CREATE COMPARISON TABLE
# ------------------------------------------------------------

results = pd.DataFrame([
    {
        "Model": rnn_result["Model"],
        "MAE": rnn_result["MAE"],
        "MSE": rnn_result["MSE"],
        "RMSE": rnn_result["RMSE"],
        "Parameters": rnn_result["Parameters"],
        "Training Time (sec)": rnn_result["Training Time (sec)"],
        "Final Train Loss": rnn_result["Final Train Loss"],
        "Final Val Loss": rnn_result["Final Val Loss"]
    },
    {
        "Model": lstm_result["Model"],
        "MAE": lstm_result["MAE"],
        "MSE": lstm_result["MSE"],
        "RMSE": lstm_result["RMSE"],
        "Parameters": lstm_result["Parameters"],
        "Training Time (sec)": lstm_result["Training Time (sec)"],
        "Final Train Loss": lstm_result["Final Train Loss"],
        "Final Val Loss": lstm_result["Final Val Loss"]
    },
    {
        "Model": gru_result["Model"],
        "MAE": gru_result["MAE"],
        "MSE": gru_result["MSE"],
        "RMSE": gru_result["RMSE"],
        "Parameters": gru_result["Parameters"],
        "Training Time (sec)": gru_result["Training Time (sec)"],
        "Final Train Loss": gru_result["Final Train Loss"],
        "Final Val Loss": gru_result["Final Val Loss"]
    }
])


# ------------------------------------------------------------
# 13. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n")
print("=" * 100)
print("Q13 - RNN vs LSTM vs GRU COMPARISON")
print("=" * 100)

print(
    results[
        [
            "Model",
            "MAE",
            "MSE",
            "RMSE",
            "Parameters",
            "Training Time (sec)",
            "Final Train Loss",
            "Final Val Loss"
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# 14. FIND BEST MODEL
# ------------------------------------------------------------

best_mae_model = results.loc[
    results["MAE"].idxmin(),
    "Model"
]

best_mse_model = results.loc[
    results["MSE"].idxmin(),
    "Model"
]

best_rmse_model = results.loc[
    results["RMSE"].idxmin(),
    "Model"
]

smallest_model = results.loc[
    results["Parameters"].idxmin(),
    "Model"
]

fastest_model = results.loc[
    results["Training Time (sec)"].idxmin(),
    "Model"
]


print("\n")
print("=" * 60)
print("BEST RESULTS")
print("=" * 60)

print("Best MAE Model       :", best_mae_model)
print("Best MSE Model       :", best_mse_model)
print("Best RMSE Model      :", best_rmse_model)
print("Fewest Parameters    :", smallest_model)
print("Fastest Training     :", fastest_model)


# ------------------------------------------------------------
# 15. PLOT TRAINING / VALIDATION LOSS
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.plot(
    rnn_result["History"].history["val_loss"],
    label="RNN Validation Loss"
)

plt.plot(
    lstm_result["History"].history["val_loss"],
    label="LSTM Validation Loss"
)

plt.plot(
    gru_result["History"].history["val_loss"],
    label="GRU Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")

plt.title(
    "RNN vs LSTM vs GRU - Validation Loss"
)

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 16. PLOT RMSE COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["RMSE"]
)

plt.xlabel("Model")
plt.ylabel("RMSE")

plt.title(
    "RNN vs LSTM vs GRU - RMSE Comparison"
)

plt.grid(axis="y")

plt.show()


# ------------------------------------------------------------
# 17. PLOT MAE COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["MAE"]
)

plt.xlabel("Model")
plt.ylabel("MAE")

plt.title(
    "RNN vs LSTM vs GRU - MAE Comparison"
)

plt.grid(axis="y")

plt.show()


# ------------------------------------------------------------
# 18. ACTUAL VS PREDICTED - ALL MODELS
# ------------------------------------------------------------

plt.figure(figsize=(14, 6))

plt.plot(
    rnn_result["Actual"],
    label="Actual"
)

plt.plot(
    rnn_result["Predicted"],
    label="Simple RNN"
)

plt.plot(
    lstm_result["Predicted"],
    label="LSTM"
)

plt.plot(
    gru_result["Predicted"],
    label="GRU"
)

plt.xlabel("Time")
plt.ylabel("Temperature")

plt.title(
    "Actual vs Predicted Temperature - RNN, LSTM and GRU"
)

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# 19. FINAL MESSAGE
# ------------------------------------------------------------

print("\n")
print("=" * 60)
print("Q13 EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 60)
```
